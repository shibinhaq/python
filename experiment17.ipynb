```python
#Read a file line by line and count the number of lines
l=[]
file1=input("Enter the source file:")
f=open(file1,"r")
for line in f.readlines():
    l.append(line)
print("Lines in the file:",len(l))
```

```python
#Copy odd lines from one file to another
file1=input("Enter the source file:")
file2=input("Enter destination file:")
f1=open(file1,"r")
f2=open(file2,"w")
i=0
for line in f1.readlines():
    i=i+1
    if i%2==0:
        continue
    f2.write(line)
f1.close()
f2.close()

```

```python
#Read each row from a CSV file
import csv
fname=input("Enter csv file name:")
f1=open(fname,"r")
reader=csv.reader(f1)
for row in reader:
    print(row)
f1.close()
```

```python
#Read specific columns from a CSV file
import csv
file1=input("Enter new filename with extension csv:")
with open(file1,"a+",newline='')as f1:
    empdata=["deptid","deptname","TE","location"]
    l=[]
    writer=csv.DictWriter(f1,fieldnames=empdata)
    writer.writeheader()
    n=int(input("Enter number of dept:"))
    for i in range(n):
        l=list(input("Enter comma seperated deptid,deptname,TE,loaction:").split(","))
        writer.writerow({'deptid':l[0],'deptname':l[1],'TE':l[2],'location':l[3]})
        l.clear()
    print("\nDepartmentID.DepartmentName")
with open(file1,"r") as f2:
    rdr=csv.DictReader(f2)
    for item in rdr:
        ##print(item['deptid'],"    ",item['deptname'])
        print(item['deptid'],"    ",item['deptname'])
```

```python
#Write a Python dictionary to a CSV file and read it
import csv
file1=input("Enter new source file name:")
empdata=["empid","empname","dptname","dob","salary"]
with open(file1,"w",newline='')as f1:
    writer=csv.DictWriter(f1,fieldnames=empdata)
    writer.writeheader()
    for i in range(2):
        l=input("Enter empid,empname,dptname,dob,salary:").split(",")
        writer.writerow({'empid':l[0],'empname':l[1],'dptname':l[2],'dob':l[3],'salary':l[4]})
print("The csv file contains:")
with open(file1,"r") as f2:
    rdr=csv.DictReader(f2)
    for item in rdr:
        print(dict(item))
    
```
