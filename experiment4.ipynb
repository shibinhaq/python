```python
#create a list of colors comma-seperated color name entered by user.display first and last color
colors=input("enter name of colors seperated by comma:").split(",")
print("first color:",colors[0])
print("last color:",colors[-1])
```

```python
#write a program to accept a filename from the user print the extention of that 
filename=input("enter filename:")
extension=filename.split(".")[-1]
print("extention:",extension)
```

```python
#program to accept a sequence of comma-seperated numbers from the user and generate the list with that number and find the sum
a=input("enter a numbers seperated by commas")
b=a.split(",")
b=[int(x) for x in b]
print("List:",b)
print("sum:",sum(b))
```
